```python
import pandas as pd
import numpy as nu
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('city_day.csv')
```

```python
data.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
l=LabelEncoder()
```

```python
data['City']=l.fit_transform(data['City'])
```

```python
data.head()
```

```python
data.isna().sum()
```

```python
data.drop("Xylene",axis=1,inplace=True)
```

```python
data.head()
```

```python
data.fillna(method="ffill",inplace=True)
```

```python
data.isna().sum()
```

```python
data['PM10'].fillna(data['PM10'].mean(),inplace=True)
```

```python
data.isna().sum()
```

```python
v={'NH3':data['NH3'].mean(),'AQI':data['AQI'].mean()}
```

```python
data.fillna(v,inplace=True)
```

```python
data.isna().sum()
```

```python
data['PM2.5'].fillna(data['PM2.5'].mean(),inplace=True)
```

```python
data.isna().sum()
```

```python
data.drop(columns=['City','Date'],inplace=True)
```

```python
data['AQI_Bucket']=l.fit_transform(data['AQI_Bucket'])
```

```python
data['AQI_Bucket'].unique()
```

```python
data.isna().sum()
```

```python
#without noise and drift
from sklearn.linear_model import LinearRegression
l=LinearRegression()
from sklearn.model_selection import train_test_split
x=data.drop(columns=["PM2.5"])
y=data['PM2.5']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2)
```

```python
l.fit(xtrain,ytrain)
ypred=l.predict(xtest)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
loss_linear=mean_absolute_error(ytest,ypred)
```

```python
loss_linear
```

```python
#neural network
import tensorflow as tf
from tensorflow.keras import models,layers
```

```python
model=models.Sequential([
    layers.Dense(10,input_shape=(12,),activation='relu'),
    layers.Dense(1,activation='sigmoid')
])
```

```python
model.compile(optimizer='adam',loss='mse',metrics=['mae'])
```

```python
model.fit(xtrain,ytrain,epochs=20)
```

```python
model.evaluate(xtest,ytest)
```
